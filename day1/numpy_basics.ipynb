```python
def dotproduct(a,b):
    #parameters are two arrays of same dimension
    
    x = 0
    for i in range(a.shape[0]):
        x = x+a[i]*b[i]
    return x


a = np.array([1, 2, 3, 4])
b = np.array([2, 2, 2, 2])

print(f"dot(a,b) = {dotproduct(a, b)}")
```

```python
r = np.dot(a, b)
print(f"dot(a,b) by numpy = {r} and shape = {r.shape}")
```
