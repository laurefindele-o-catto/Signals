```python
class Cat:
    def __init__(self, name, floof):
        self.name = name
        self.floof = floof
        
    def sleep(self):
        return f'{self.name} sleeps'
    
    def __str__(self):
        return f'{self.name} is a {self.floof} on the floofy scale.'
```

```python
tom = Cat('tom', 9)
print(tom)
```
