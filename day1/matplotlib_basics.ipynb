```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
xs = np.array([-1, 0, 2, 3, 10])
ys = np.array([1, 9, 14, 5, 20])

plt.plot(xs, ys, marker = 'o', color = 'blue', linestyle = '--')
plt.show()
```

```python
xs = np.array([-1, 0, 2, 3, 10])
ys = np.array([1, 9, 14, 5, 20])

plt.plot(xs, ys, marker = 'o', color = 'blue', linestyle = '--')

plt.subplot(2,1,2)
xs = np.array([-5, 1, 4, 6, 15])
ys = np.array([-10, 4, 0, 10, 30])

plt.plot(xs,ys, marker = 'o', color='hotpink', linestyle = '--')

plt.show()
```

```python
sampling_f = 1000

t = np.linspace(0, 1, sampling_f)
f = 5
y = np.sin(2*np.pi*f*t)

plt.figure(1)
plt.plot(t,y)
plt.grid(True)
plt.show()

square_wave = np.sign(y)
plt.figure(2)
plt.plot(t,square_wave)
plt.show()
```

```python
plt.stem(t, y)
plt.show()
```
